# Andy's Afternoon Amble

## Observations

### Soccer ball

* The soccer ball can be represented by the graph $K_4$, where one of the vertices is the home vertex.
* A path of length $k \geq 2$ consists of going to an arbitrary vertex, taking $k - 2$ steps among the non-home vertices, and coming back.
* Traversing the non-home vertices can be modeled by traversing the cycle graph $C_3$.
* The probability of taking a path of length $k \geq 2$ is thus $P_k = \frac{1}{3} \cdot \left(\frac{2}{3}\right)^{k-2}$.
* What does such a path look like? Denote the directions as L = left, R = right, B = back. First, Andy goes L. Then he chooses either L or R. If he chooses R, then at each subsequent step, he cannot choose R. If he chooses L, the situation repeats. If he chooses B, then the situation rotates and he cannot choose L, since this would take him home early.

### Kitchen floor

* Denote the home hexagon by $H$.
* At the start of his walk, he goes to some white hexagon $S$. $S$ and $H$ share exactly $2$ black hexagons.
* If Andy goes in a circle around one of these hexagons, he will reach home on the soccer ball on his third step, because he will perform an R,R move sequence, which leads him home on the soccer ball.
* Thus, every path on the kitchen floor that is indistinguishable from the soccer ball path goes to $S$, goes around the black hexagon it doesn't share with $H$, and finally goes back.

## Solution

We can model this as a graph traversal, which can be simulated by computing powers of the adjacency matrix. Every path on the kitchen floor that is indistinguishable from the soccer ball path consists of an initial step to $S$, some number of steps around the black hexagon (modeled by traversing $C_6$), and a final step back to $H$. The probability that Andy won't discover he isn't on the soccer ball is thus the sum of the probabilities of all paths that start at $S$, end at $S$, and never leave $C_6$.

```python
import numpy as np

ITERS = 1000

# neighboring matrix with probabilities.
# The missing 1/3 probability is ending early, e.g. ending at soccer ball but not kitchen floor
A = np.array([
    [  0, 1/3,   0,   0,   0, 1/3],
    [1/3,   0, 1/3,   0,   0,   0],
    [  0, 1/3,   0, 1/3,   0,   0],
    [  0,   0, 1/3,   0, 1/3,   0],
    [  0,   0,   0, 1/3,   0, 1/3],
    [1/3,   0,   0,   0, 1/3,   0],
])

ans = 0.
cur_A = np.eye(A.shape[0])

for _ in range(ITERS):
    ans += 1/3 * cur_A[0, 0]
    cur_A = cur_A @ A

print(1 - ans)
```

The answer 0.55 is equivalent to $11/20$, which is the answer
